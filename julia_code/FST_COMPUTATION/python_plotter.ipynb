```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import glob

# Set the global font to Times New Roman
plt.rcParams['font.family'] = 'serif'
plt.rcParams['font.serif'] = ['Times New Roman']
plt.rcParams['font.size'] = 12  # Default font size
plt.rcParams['axes.labelsize'] = 12
plt.rcParams['axes.titlesize'] = 12
plt.rcParams['xtick.labelsize'] = 12
plt.rcParams['ytick.labelsize'] = 12
plt.rcParams['legend.fontsize'] = 9
written_text_fontsize = 12

plt.rcParams['figure.dpi'] = 100  # or set it to any constant value

files = glob.glob("FST_m*.csv")
# Extract numeric migration rates and sort by them
files_and_rates = [(f, float(f.split("FST_m")[1].split(".csv")[0])) for f in files]
files_and_rates.sort(key=lambda x: x[1])  # sort by migration rate

csv_files = [f for f, _ in files_and_rates]
migration_rates = [r for _, r in files_and_rates]

# Color-blind safe palette
colors = plt.cm.plasma(np.linspace(0, 1, len(csv_files)))

# Plot
fig, ax = plt.subplots(figsize=(6.5/2, 2.75))
for i, fname in enumerate(csv_files):
    df = pd.read_csv(fname)
    time = df['time']
    mean = df['FST']
    std = df['FST_error']

    # ax.plot(time, mean, label=f"${migration_rates[i]}$", color=colors[i])
    exp = int(np.log10(migration_rates[i]))
    ax.plot(time, mean, label=rf"$10^{{{exp}}}$", color=colors[i], lw=1)

    ax.fill_between(time, mean - std/np.sqrt(200), mean + std/np.sqrt(200), color=colors[i], alpha=0.25)

ax.set_xlabel(r"time, $t$")
ax.set_ylabel(r"fixation index, $F_{ST}$")
ax.set_xlim(0, 20)
ax.set_ylim(0.9, 1.01)
ax.legend(title="migration rate", loc=(0.05, 0.01), fontsize=7, title_fontsize=7)
plt.tight_layout()
plt.savefig("fixation_index.pdf")
plt.show()
```
