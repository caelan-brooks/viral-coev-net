```julia
using CSV
using DataFrames
using Serialization
# using HDF5
using Statistics
using Plots
```

```julia
df = DataFrame(CSV.File("../real_network/cleaned_adjacency_matrix.csv"));
```

```julia
real_migration_matrix = Matrix(df)[:,1:end-1]
```

```julia
p = plot()
for i = 1:30
    for j = i+1:30
        scatter!(p, [log10(real_migration_matrix[i, j])], [log10(real_migration_matrix[j, i])], color=:black, legend=false)
    end
end
plot!(p, -10:0, -10:0)
display(p)
```

```julia
migration_rates = []
for i = 1:30
    for j=1:30
        if i==j
            continue
        else
            push!(migration_rates, real_migration_matrix[i,j])
        end
    end
end
```

```julia
barhist(log10.(migration_rates))
```

```julia
bootstrapped_migration_matrix = fill(NaN, 30, 30)
```

```julia
for i=1:30
    for j=i:30
        if i==j 
            bootstrapped_migration_matrix[i,j] = 0.0
        else
            bootstrapped_migration_matrix[i,j] = migration_rates[rand(1:length(migration_rates))]
            bootstrapped_migration_matrix[j,i] = bootstrapped_migration_matrix[i,j]
        end
    end
end
```

```julia
bootstrapped_migration_matrix
```

```julia
bootstrapped_migration_rates = []
for i = 1:30
    for j=1:30
        if i==j
            continue
        else
            push!(bootstrapped_migration_rates, bootstrapped_migration_matrix[i,j])
        end
    end
end
```

```julia
barhist!(log10.(bootstrapped_migration_rates))
```

```julia
p = plot()
for i = 1:30
    for j = i+1:30
        scatter!(p, [log10(bootstrapped_migration_matrix[i, j])], [log10(bootstrapped_migration_matrix[j, i])], color=:black, legend=false)
    end
end
plot!(p, -10:0, -10:0)
display(p)
```

```julia
df = DataFrame(bootstrapped_migration_matrix,:auto)
CSV.write("bootstrapped_migration_matrix.csv", df)
```
